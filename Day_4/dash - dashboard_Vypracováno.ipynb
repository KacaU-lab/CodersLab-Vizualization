```python
import dash
import pandas as pd
import numpy as np
import plotly.express as px
import threading
from dash import dcc
from dash import html
from dash.dependencies import Input, Output
from google.colab import output
```

### Creating the visualization

#### Supporting solutions

We will use the following function to create a simple visualization.

```python
def find_special_points(value, points):

    if value == points.max():
      special_point_name = "max"
    elif value == points.min():
      special_point_name = "min"
    elif value == points.median():
      special_point_name ="median"
    else:
      special_point_name = "not special"

    return special_point_name
```

Below, we have the full code of the application used in the **Dashboard** presentation.

```python
app = dash.Dash()
app.layout = html.Div(
        children=[
         html.Div(children=[
                            html.H2("Random points generator"),
                            html.H4("This dashboard generates random points and highlights max, min and median based on the value you select.")
         ]),
       html.Div(
            children=[
                dcc.Graph(id="graph"),
                html.Div(
                    children=[
                        html.P("Choose value"),
                        dcc.Dropdown(
                            id="number-dropdown",
                            clearable=False,
                            value=7,
                        ),
                    ],
                    style={
                        "display": "flex",
                        "flex-direction": "row",
                        "justify-content": "center",
                        "align-items": "center",
                        "height": "100%"
                    }
                ),
            ],
            style={
                "display": "flex",
                "flex-direction": "column",
                "justify-content": "space-around",
                "height": "100%",
                "fontFamily": "Verdana",
                "color": "#4044"
            }
        )
         ],
         style={
                "display": "flex",
                "flex-direction": "row",      # ⬅️ ZMĚNA: column -> row
                "justify-content": "space-around",
                "align-items": "stretch",     # ⬅️ NOVÉ – prvky vyplní výšku řádku
                "height": "100%",
                "fontFamily": "Verdana",
                "color": "#404040"
            }
)

@app.callback(
    Output('graph', 'figure'),
    [Input("number-dropdown", "value")]
)
def update_figure(points_number):
    numbers = pd.DataFrame(np.random.randn(points_number), columns = ["sample",])
    numbers["special_points"]=numbers["sample"].apply(lambda x: find_special_points(x,numbers["sample"]))

    chart = px.scatter(
            numbers,
            x=numbers.index,
            y="sample",
            color="special_points",
            color_discrete_map={
                "max": "#800E13",
                "median": "#169873",
                "min": "#2C497F",
                "not special": "#E4DFDA",
                    },
            render_mode="webgl",
            template='none',
            size=[10 for item in numbers["sample"]],
            title="Your points"
        )
    chart
    return chart
```

### Displaying the visualization

To build the visualization using the code above, we also need to run a server and go to the application page.


```python
thread = threading.Thread(
    target=app.run,
    kwargs={"port": 8050, "debug": False, "use_reloader": False}
)
thread.start()
output.serve_kernel_port_as_window(port=8050)
```

```python
app = dash.Dash()

app.layout = html.Div(
    children=[

        # HLAVIČKA
        html.Div(
            children=[
                html.H2("Random points generator"),
                html.H4(
                    "This dashboard generates random points and highlights max, min and median based on the value you select."
                ),
            ],
            style={
                "textAlign": "center",
                "marginBottom": "30px"
            }
        ),

        # HLAVNÍ KONTEJNER – graf + panel pod sebou
        html.Div(
            children=[

                # GRAF
                dcc.Graph(id="graph"),

                # VNITŘNÍ KONTEJNER – label + dropdown v řádku
                html.Div(
                    children=[
                        html.P("Choose value", style={"margin-right": "10px"}),

                        dcc.Dropdown(
                            id="number-dropdown",
                            options=[{"label": str(i), "value": i} for i in range(1, 11)],
                            clearable=False,
                            value=7,
                            style={"width": "120px"}
                        ),
                    ],
                    style={
                        "display": "flex",
                        "flex-direction": "row",
                        "justify-content": "center",
                        "align-items": "center",
                        "margin-top": "20px",
                    }
                ),
            ],
            style={
                "display": "flex",
                "flex-direction": "column",   # prvky pod sebou
                "justify-content": "center",
                "align-items": "center",
                "width": "100%",
                "fontFamily": "Verdana",
            }
        ),
    ]
)


@app.callback(
    Output("graph", "figure"),
    [Input("number-dropdown", "value")]
)
def update_figure(points_number):
    numbers = pd.DataFrame(np.random.randn(points_number), columns=["sample"])
    numbers["special_points"] = numbers["sample"].apply(
        lambda x: find_special_points(x, numbers["sample"])
    )

    chart = px.scatter(
        numbers,
        x=numbers.index,
        y="sample",
        color="special_points",
        color_discrete_map={
            "max": "#800E13",
            "median": "#169873",
            "min": "#2C497F",
            "not special": "#E4DFDA",
        },
        render_mode="webgl",
        template="none",
        size=[10 for _ in numbers["sample"]],
        title="Your points",
    )
    return chart
```

```python
from google.colab import output

# vybereme si port, který nejspíš není obsazený
PORT = 8060

thread = threading.Thread(
    target=app.run,
    kwargs={"port": PORT, "debug": False, "use_reloader": False}
)
thread.start()

# vykreslení appky přímo do buňky jako iframe
output.serve_kernel_port_as_iframe(port=PORT, height=600)

print("\n--- Souhrn ---")
print(f"1) Dash aplikace běží na portu {PORT}.")
print("2) Nad tímto textem bys měla vidět vložené okno (iframe) s dashboardem.")
print("3) Pokud iframe nevidíš, zkus spustit buňku znovu nebo obnovit stránku.")
```
