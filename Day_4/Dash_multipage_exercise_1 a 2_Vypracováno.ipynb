```python
import pandas as pd
import numpy as np
import plotly.express as px

from dash import Dash, dcc, html, Input, Output
import threading
from google.colab import output

# ---------------------------------------------------
# 1) Načtení dat – uprav si base_path podle svého disku
# ---------------------------------------------------
base_path = "/content/drive/MyDrive/Colab Notebooks/Vis/"

population = pd.read_csv(base_path + "population.csv")
new_books = pd.read_csv(base_path + "new-books-per-million.csv")

selected_books = new_books[(new_books["Year"] > 1940) & (new_books["Year"] <= 1996)]
selected_pops = population[population["Year"] == 1996][["Entity", "Population (historical estimates)"]]

data = pd.merge(selected_books, selected_pops, how="inner", on="Entity")

# ---------------------------------------------------
# 2) Graf pro druhou stránku (scatter + line)
# ---------------------------------------------------
scatter_figure = px.scatter(
    data_frame=data,
    x="Book titles per capita (Fink-Jensen 2015)",
    y="Population (historical estimates)",
    text="Entity",
    log_x=True,
    log_y=True,
    template="none",
    title="Country population vs number of new book titles per capita in 1996",
)
scatter_figure.update_traces(textposition="bottom right")

# ---------------------------------------------------
# 3) Layout druhé stránky – beze změny
# ---------------------------------------------------
layout_page_two = html.Div(
    children=[
        dcc.Graph(
            id="scatter",
            figure=scatter_figure,
        ),
        dcc.Graph(id="line"),
    ]
)

# ---------------------------------------------------
# 4) Layout první stránky – ÚPRAVA PODLE ZADÁNÍ
#    místo dropdownu je textový input
# ---------------------------------------------------
layout_page_one = html.Div(
    children=[
        dcc.Graph(id="some-lines"),
        html.Br(),
        html.P("Zadej název země v angličtině (musí odpovídat sloupci 'Entity')"),
        dcc.Input(
            id="country-input",
            type="text",
            placeholder="např. Switzerland",
            value="Switzerland",
            style={"width": "300px"},
        ),
        html.P(id="country-help", style={"marginTop": "10px", "color": "#666"}),
    ]
)

# ---------------------------------------------------
# 5) Vlastní aplikace – multipage layout
# ---------------------------------------------------
app = Dash()

app.layout = html.Div(
    children=[
        dcc.Location(id="url", refresh=False),

        dcc.Link("Linechart with typed country name", href="/"),
        html.Br(),
        dcc.Link("Cross filtering in dash", href="/second_page"),
        html.Hr(),

        html.Div(id="page-content"),
    ]
)

# ---------------------------------------------------
# 6) Callback pro přepínání stránek
# ---------------------------------------------------
@app.callback(
    Output("page-content", "children"),
    Input("url", "pathname"),
)
def display_page(pathname):
    if pathname == "/second_page":
        return layout_page_two
    else:
        return layout_page_one

# ---------------------------------------------------
# 7) Callback pro cross-filtering (druhá stránka)
#     beze změny – jen čte hoverData ze scatteru
# ---------------------------------------------------
@app.callback(
    Output("line", "figure"),
    Input("scatter", "hoverData"),
)
def build_line_chart(country):
    if country is None:
        # nic ještě nebylo vybráno – vrátíme prázdný graf
        fig = px.line(template="none")
        fig.update_layout(title="Move the mouse over a point in the scatter plot.")
        return fig

    selected_country = country["points"][0]["text"]
    selected_data = new_books[new_books["Entity"] == selected_country]

    fig = px.line(
        data_frame=selected_data,
        x="Year",
        y="Book titles per capita (Fink-Jensen 2015)",
        template="none",
        title="Book titles per capita in kx for years 1948–1996 – " + selected_country,
    )
    return fig

# ---------------------------------------------------
# 8) NOVÝ callback pro první stránku – textový input
# ---------------------------------------------------
@app.callback(
    Output("some-lines", "figure"),
    Output("country-help", "children"),
    Input("country-input", "value"),
)
def build_country_line(country_name):
    # ošetření prázdného vstupu
    if not country_name:
        fig = px.line(template="none")
        fig.update_layout(title="Please type a country name in English.")
        msg = "Zatím nebyl zadán žádný název země."
        return fig, msg

    # case-insensitive hledání v 'Entity'
    mask = new_books["Entity"].str.lower() == country_name.strip().lower()

    if not mask.any():
        # žádná shoda – vrátíme prázdný graf + info text
        fig = px.line(template="none")
        fig.update_layout(
            title=f"No data for '{country_name}'. Please type a country name exactly as in the dataset."
        )
        msg = f"Nebyla nalezena země „{country_name}“. Zkontroluj prosím anglický název."
        return fig, msg

    # máme shodu – vezmeme první (správně zapsaný) název
    present_books = new_books[mask]
    proper_name = present_books["Entity"].iloc[0]

    fig = px.line(
        data_frame=present_books,
        x="Year",
        y="Book titles per capita (Fink-Jensen 2015)",
        template="none",
        title=f"Book titles per capita in {proper_name} for years 1948–1996",
    )

    msg = f"Zobrazuji data pro zemi: {proper_name}."
    return fig, msg

# ---------------------------------------------------
# 9) Spuštění aplikace v Colabu – thread + iframe + localhost
# ---------------------------------------------------
PORT = 8050

def run_app():
    app.run(
        host="0.0.0.0",
        port=PORT,
        debug=False,
        use_reloader=False,
    )

thread = threading.Thread(target=run_app)
thread.start()

output.serve_kernel_port_as_iframe(port=PORT, height=600)

print("\n--- Souhrn ---")
print(f"1) Dash multipage aplikace běží na portu {PORT}.")
print(f"2) Odkaz localhost (mimo Colab prostředí): http://127.0.0.1:{PORT}/")
print("3) V tomto notebooku vidíš aplikaci nad tímto textem v iframe.")
print("4) Pokud iframe nevidíš, spusť buňku znovu nebo obnov stránku.")
```

```python
# ==== 1) Importy a připojení Google Drive ====================================
from google.colab import drive, output
drive.mount('/content/drive')

import os
import threading
import pandas as pd
import plotly.express as px
from dash import Dash, dcc, html, Input, Output, callback_context

# ==== 2) Cesty k souborům a načtení dat ======================================
base_path = "/content/drive/MyDrive/Colab Notebooks/Vis"
population_path = os.path.join(base_path, "population.csv")
new_books_path = os.path.join(base_path, "new-books-per-million.csv")

population = pd.read_csv(population_path)
new_books = pd.read_csv(new_books_path)

# pro line grafy – jen roky 1948–1996
new_books = new_books[(new_books["Year"] >= 1948) & (new_books["Year"] <= 1996)]

# ---- data pro scatter (page 2) ----------------------------------------------
selected_books = new_books[new_books["Year"] == 1996][[
    "Entity", "Book titles per capita (Fink-Jensen 2015)"
]]
selected_pops = population[population["Year"] == 1996][[
    "Entity", "Population (historical estimates)"
]]
data = pd.merge(selected_books, selected_pops, how="inner", on="Entity")

# ---- data pro ranking populace (page 3) -------------------------------------
# odfiltrujeme agregované jednotky (svět / kontinenty / EU apod.)
aggregates = [
    "World", "Africa", "Asia", "Europe",
    "North America", "South America", "Oceania",
    "European Union (27)", "European Union"
]

selected_pops_countries = (
    selected_pops[~selected_pops["Entity"].isin(aggregates)]
    .sort_values("Population (historical estimates)", ascending=False)
)

# ==== 3) Scatter figure pro druhou stránku ===================================
scatter_figure = px.scatter(
    data_frame=data,
    x="Book titles per capita (Fink-Jensen 2015)",
    y="Population (historical estimates)",
    text="Entity",
    log_x=True,
    log_y=True,
    template="none",
    title="Country population vs number of new book titles per capita in 1996",
)
scatter_figure.update_traces(textposition="bottom right")


# ==== 4) Layout první stránky (input + dropdown) =============================
layout_page_one = html.Div(
    children=[
        dcc.Graph(id="some-lines"),

        html.Br(),
        html.H4("Vyber zemi nebo ji napiš ručně"),

        html.Label("1) Výběr ze seznamu (dropdown)"),
        dcc.Dropdown(
            id="country-dropdown",
            options=sorted(new_books["Entity"].unique()),
            value="Switzerland",
            placeholder="Vyber zemi...",
            style={"width": "300px"},
        ),

        html.Br(),

        html.Label("2) Nebo napiš název země v angličtině"),
        dcc.Input(
            id="country-input",
            type="text",
            placeholder="např. Switzerland",
            value="Switzerland",
            style={"width": "300px"},
        ),

        html.P(id="country-help", style={"marginTop": "10px", "color": "#666"}),
    ]
)

# ==== 5) Layout druhé stránky (cross filtering) ==============================
layout_page_two = html.Div(
    children=[
        dcc.Graph(id="scatter", figure=scatter_figure),
        dcc.Graph(id="line"),
    ]
)

# ==== 6) Layout třetí stránky (ranking populace + slider) =====================
layout_page_three = html.Div(
    children=[
        html.H3("Top countries by population in 1996"),
        html.P(
            "Posuvníkem zvol počet zemí v grafu (zobrazují se jen jednotlivé státy, "
            "ne agregované hodnoty pro svět/kontinenty)."
        ),
        dcc.Slider(
            id="country-count-slider",
            min=5,
            max=30,
            step=5,
            value=10,
            marks={i: str(i) for i in range(5, 31, 5)},
        ),
        html.Br(),
        html.Div(id="slider-help", style={"color": "#666"}),
        dcc.Graph(id="population-bar"),
    ]
)


# ==== 7) Dash aplikace – multipage layout ===================================
app = Dash()

app.layout = html.Div([
    dcc.Location(id="url", refresh=False),

    dcc.Link("Linechart with typed/selected country", href="/"),
    html.Br(),
    dcc.Link("Cross filtering in dash", href="/second_page"),
    html.Br(),
    dcc.Link("Population ranking", href="/third_page"),
    html.Hr(),

    html.Div(id="page-content")
])


# ---- Přepínání stránek podle URL -------------------------------------------
@app.callback(
    Output("page-content", "children"),
    Input("url", "pathname"),
)
def display_page(pathname):
    if pathname == "/second_page":
        return layout_page_two
    elif pathname == "/third_page":
        return layout_page_three
    return layout_page_one


# ---- Callback pro první stránku: input + dropdown ---------------------------
@app.callback(
    Output("some-lines", "figure"),
    Output("country-help", "children"),
    Input("country-input", "value"),
    Input("country-dropdown", "value"),
)
def build_country_line(country_input, country_dropdown):
    ctx = callback_context
    if ctx.triggered:
        triggered_id = ctx.triggered[0]["prop_id"].split(".")[0]
    else:
        triggered_id = None

    if triggered_id == "country-dropdown":
        chosen = country_dropdown
        source = "dropdownu"
    elif triggered_id == "country-input":
        chosen = country_input
        source = "textového pole"
    else:
        chosen = country_dropdown or country_input
        source = "výchozí hodnoty"

    if not chosen:
        fig = px.line(template="none")
        fig.update_layout(
            title="Vyber zemi v dropdownu nebo ji napiš do textového pole."
        )
        msg = "Zatím nebyl zadán ani vybrán žádný název země."
        return fig, msg

    mask = new_books["Entity"].str.lower() == chosen.strip().lower()

    if not mask.any():
        fig = px.line(template="none")
        fig.update_layout(
            title=f"No data for '{chosen}'. Please type a country name exactly as in the dataset."
        )
        msg = (
            f"Nenalezena země „{chosen}“. Zkus jiný název nebo použij dropdown."
        )
        return fig, msg

    present_books = new_books[mask]
    proper_name = present_books["Entity"].iloc[0]

    fig = px.line(
        data_frame=present_books,
        x="Year",
        y="Book titles per capita (Fink-Jensen 2015)",
        template="none",
        title=f"Book titles per capita in {proper_name} for years 1948–1996",
    )

    msg = f"Zobrazuji data pro zemi: {proper_name} (vybráno z {source})."
    return fig, msg


# ---- Callback pro druhou stránku: cross filtering ---------------------------
@app.callback(
    Output("line", "figure"),
    Input("scatter", "hoverData"),
)
def build_line_chart(country):
    if country is None:
        default_country = new_books["Entity"].iloc[0]
        selected_data = new_books[new_books["Entity"] == default_country]
    else:
        selected_country = country["points"][0]["text"]
        selected_data = new_books[new_books["Entity"] == selected_country]
        default_country = selected_country

    fig = px.line(
        data_frame=selected_data,
        x="Year",
        y="Book titles per capita (Fink-Jensen 2015)",
        template="none",
        title=f"Book titles per capita in {default_country} for years 1948–1996",
    )
    return fig


# ---- Callback pro třetí stránku: bar graf + slider --------------------------
@app.callback(
    Output("population-bar", "figure"),
    Output("slider-help", "children"),
    Input("country-count-slider", "value"),
)
def update_population_bar(count):
    if count is None:
        count = 10
    n = int(count)

    top_countries = selected_pops_countries.head(n)

    fig = px.bar(
        top_countries,
        x="Entity",
        y="Population (historical estimates)",
        template="none",
        title=f"Top {n} countries by population in 1996",
    )
    fig.update_layout(
        xaxis_title="Country",
        yaxis_title="Population (historical estimates)",
        xaxis_tickangle=-45,
    )

    msg = f"Zobrazuji {n} zemí podle populace v roce 1996."
    return fig, msg


# ==== 8) Spuštění serveru v samostatném vlákně + iframe ======================
PORT = 8050

thread = threading.Thread(
    target=app.run,
    kwargs={"port": PORT, "debug": False, "use_reloader": False},
)
thread.start()

output.serve_kernel_port_as_iframe(port=PORT, height=600)

print("\n--- Souhrn ---")
print(f"1) Dash multipage aplikace běží na portu {PORT}.")
print(f"2) Odkaz localhost (mimo Colab prostředí): http://127.0.0.1:{PORT}/")
print("3) V tomto notebooku vidíš aplikaci nad tímto textem v iframe.")
print("4) Pokud iframe nevidíš, spusť buňku znovu nebo obnov stránku.")
```

V tomto cvičení jsme rozšířili vícestránkovou Dash aplikaci o třetí podstránku, která vizualizuje pořadí zemí podle populace v roce 1996. Úkol byl zaměřený na dvě hlavní dovednosti:

Umět filtrovat datový rámec podle zadání (vyloučit agregované oblasti – kontinenty, svět apod.).

Použít hodnotu posuvníku jako vstup do callbacku a vracet dynamicky vytvořený graf.

Na nové podstránce se řeší:

1️⃣ Sloupcový graf – pořadí zemí podle počtu obyvatel (1996)

Data byla nejdříve očištěna od agregovaných entit (World, Europe, Asia…).

Z filtrů tedy zůstaly pouze jednotlivé země, aby řazení podle populace dávalo smysl.

Tabulka byla seřazena sestupně a pomocí .head(N) se zobrazí pouze počet zemí zvolený v posuvníku.

2️⃣ Posuvník pro volbu počtu zemí (5–30)

Slider vrací celé číslo (5, 10, 15, … 30).

Tato hodnota je předána callbacku, který ji použije jako parametr metody .head(N).

Uživatel tak může interaktivně měnit množství zobrazených zemí přímo v grafu.

3️⃣ Callback – propojení slideru a grafu

Callback má 1 vstup (slider) a 2 výstupy (graf + textová zpráva).

Funguje to tak, že při každé změně posuvníku se:

znovu vybere prvních N zemí podle populace,

vygeneruje odpovídající bar chart,

zobrazí se krátká informace, kolik zemí právě vidíme.

💡 Hlavní myšlenka cvičení

Cílem bylo ukázat, jak může jednoduchá numerická hodnota z komponenty (slider) dynamicky měnit rozsah dat v grafu. Jde o jeden z nejběžnějších způsobů interaktivity v aplikacích Dash.

```python
# ================================================================
# 1) Připojení Google Drivu + import potřebných knihoven
# ================================================================
from google.colab import drive, output
drive.mount('/content/drive')   # připojení tvého Google Drivu do /content/drive

import os
import threading
import pandas as pd
import plotly.express as px
from dash import Dash, dcc, html, Input, Output, callback_context


# ================================================================
# 2) Cesty k souborům a načtení dat
#    – používáme stejné soubory jako v předchozích cvičeních
# ================================================================
base_path = "/content/drive/MyDrive/Colab Notebooks/Vis"
population_path = os.path.join(base_path, "population.csv")
new_books_path = os.path.join(base_path, "new-books-per-million.csv")

# načtení CSV souborů
population = pd.read_csv(population_path)
new_books = pd.read_csv(new_books_path)

# pro line grafy použijeme jen roky 1948–1996
new_books = new_books[(new_books["Year"] >= 1948) & (new_books["Year"] <= 1996)]

# ------------------------------------------------
# data pro scatter (2. stránka – cross filtering)
# ------------------------------------------------
selected_books = new_books[new_books["Year"] == 1996][[
    "Entity", "Book titles per capita (Fink-Jensen 2015)"
]]
selected_pops = population[population["Year"] == 1996][[
    "Entity", "Population (historical estimates)"
]]

# spojení obou tabulek podle názvu Entity (země)
data = pd.merge(selected_books, selected_pops, how="inner", on="Entity")

# ------------------------------------------------
# data pro ranking populace (3. stránka)
# – musíme odfiltrovat agregované oblasti (svět, kontinenty…)
# ------------------------------------------------
aggregates = [
    "World", "Africa", "Asia", "Europe",
    "North America", "South America", "Oceania",
    "European Union (27)", "European Union"
]

selected_pops_countries = (
    selected_pops[~selected_pops["Entity"].isin(aggregates)]
    .sort_values("Population (historical estimates)", ascending=False)
)


# ================================================================
# 3) Scatter graf pro stránku 2 (cross filtering)
# ================================================================
scatter_figure = px.scatter(
    data_frame=data,
    x="Book titles per capita (Fink-Jensen 2015)",
    y="Population (historical estimates)",
    text="Entity",                # popisek bodu = název země
    log_x=True,                   # logaritmické měřítko osy X
    log_y=True,                   # logaritmické měřítko osy Y
    template="none",
    title="Country population vs number of new book titles per capita in 1996",
)
scatter_figure.update_traces(textposition="bottom right")


# ================================================================
# 4) Layout první stránky – line chart + input + dropdown
# ================================================================
layout_page_one = html.Div(
    children=[
        dcc.Graph(id="some-lines"),   # sem se bude kreslit line graf

        html.Br(),
        html.H4("Vyber zemi nebo ji napiš ručně"),

        # Dropdown – výběr ze seznamu zemí
        html.Label("1) Výběr ze seznamu (dropdown)"),
        dcc.Dropdown(
            id="country-dropdown",
            options=sorted(new_books["Entity"].unique()),
            value="Switzerland",
            placeholder="Vyber zemi...",
            style={"width": "300px"},
        ),

        html.Br(),

        # Textový input – ruční zadání země
        html.Label("2) Nebo napiš název země v angličtině"),
        dcc.Input(
            id="country-input",
            type="text",
            placeholder="např. Switzerland",
            value="Switzerland",
            style={"width": "300px"},
        ),

        # krátká nápověda / status pod vstupy
        html.P(id="country-help", style={"marginTop": "10px", "color": "#666"}),
    ]
)


# ================================================================
# 5) Layout druhé stránky – cross filtering
#    scatter (1996) + line graf podle zvolené země
# ================================================================
layout_page_two = html.Div(
    children=[
        dcc.Graph(id="scatter", figure=scatter_figure),
        dcc.Graph(id="line"),
    ]
)


# ================================================================
# 6) Layout třetí stránky – bar graf + slider (ranking populace)
# ================================================================
layout_page_three = html.Div(
    children=[
        html.H3("Top countries by population in 1996"),
        html.P(
            "Posuvníkem zvol počet zemí v grafu (zobrazují se jen jednotlivé státy, "
            "ne agregované hodnoty pro svět/kontinenty)."
        ),

        # Slider – kolik zemí se má zobrazit (min 5, max 30, krok 5)
        dcc.Slider(
            id="country-count-slider",
            min=5,
            max=30,
            step=5,
            value=10,
            marks={i: str(i) for i in range(5, 31, 5)},
        ),
        html.Br(),

        # textová informace pod sliderem
        html.Div(id="slider-help", style={"color": "#666"}),

        # bar graf s rankingem
        dcc.Graph(id="population-bar"),
    ]
)


# ================================================================
# 7) Vytvoření Dash aplikace + multipage layout
# ================================================================
app = Dash()

app.layout = html.Div([
    # Location sleduje URL a umožňuje přepínat stránky
    dcc.Location(id="url", refresh=False),

    # jednoduché menu odkazů na jednotlivé stránky
    dcc.Link("Linechart with typed/selected country", href="/"),
    html.Br(),
    dcc.Link("Cross filtering in dash", href="/second_page"),
    html.Br(),
    dcc.Link("Population ranking", href="/third_page"),
    html.Hr(),

    # sem se bude vkládat obsah aktuální stránky
    html.Div(id="page-content")
])


# ================================================================
# 8) Callback pro přepínání stránek podle URL
# ================================================================
@app.callback(
    Output("page-content", "children"),
    Input("url", "pathname"),
)
def display_page(pathname):
    """Vrací layout podle toho, jaká je aktuální URL."""
    if pathname == "/second_page":
        return layout_page_two
    elif pathname == "/third_page":
        return layout_page_three
    # default = první stránka
    return layout_page_one


# ================================================================
# 9) Callback pro první stránku – line chart + input + dropdown
# ================================================================
@app.callback(
    Output("some-lines", "figure"),
    Output("country-help", "children"),
    Input("country-input", "value"),
    Input("country-dropdown", "value"),
)
def build_country_line(country_input, country_dropdown):
    """
    Zvolí zemi buď podle toho, co bylo naposledy změněno
    (input vs dropdown), nebo použije výchozí hodnotu.
    Potom vykreslí line graf pro zvolenou zemi.
    """
    # zjistíme, který vstup callback spustil
    ctx = callback_context
    if ctx.triggered:
        triggered_id = ctx.triggered[0]["prop_id"].split(".")[0]
    else:
        triggered_id = None

    # rozhodnutí, odkud vzít název země
    if triggered_id == "country-dropdown":
        chosen = country_dropdown
        source = "dropdownu"
    elif triggered_id == "country-input":
        chosen = country_input
        source = "textového pole"
    else:
        # první vykreslení – vezmeme cokoliv, co je k dispozici
        chosen = country_dropdown or country_input
        source = "výchozí hodnoty"

    # nic nezadáno ani nevybráno
    if not chosen:
        fig = px.line(template="none")
        fig.update_layout(
            title="Vyber zemi v dropdownu nebo ji napiš do textového pole."
        )
        msg = "Zatím nebyl zadán ani vybrán žádný název země."
        return fig, msg

    # porovnáváme case-insensitive
    mask = new_books["Entity"].str.lower() == chosen.strip().lower()

    # neplatný název země – zobrazíme jen textovou informaci
    if not mask.any():
        fig = px.line(template="none")
        fig.update_layout(
            title=f"No data for '{chosen}'. Please type a country name exactly as in the dataset."
        )
        msg = f"Nenalezena země „{chosen}“. Zkus jiný název nebo použij dropdown."
        return fig, msg

    # vyfiltrujeme data pro danou zemi
    present_books = new_books[mask]
    proper_name = present_books["Entity"].iloc[0]   # „čistý“ název ze souboru

    fig = px.line(
        data_frame=present_books,
        x="Year",
        y="Book titles per capita (Fink-Jensen 2015)",
        template="none",
        title=f"Book titles per capita in {proper_name} for years 1948–1996",
    )

    msg = f"Zobrazuji data pro zemi: {proper_name} (vybráno z {source})."
    return fig, msg


# ================================================================
# 10) Callback pro druhou stránku – cross filtering
#      (hovernad bodem -> line graf pro zvolenou zemi)
# ================================================================
@app.callback(
    Output("line", "figure"),
    Input("scatter", "hoverData"),
)
def build_line_chart(country):
    """
    Podle bodu, nad kterým je kurzor ve scatter grafu,
    vybere zemi a zobrazí její časovou řadu v line grafu.
    Pokud ještě nikdo nenajel, zobrazí první zemi v datasetu.
    """
    if country is None:
        # default – když ještě nikdo nenajel myší
        default_country = new_books["Entity"].iloc[0]
        selected_data = new_books[new_books["Entity"] == default_country]
    else:
        selected_country = country["points"][0]["text"]
        selected_data = new_books[new_books["Entity"] == selected_country]
        default_country = selected_country

    fig = px.line(
        data_frame=selected_data,
        x="Year",
        y="Book titles per capita (Fink-Jensen 2015)",
        template="none",
        title=f"Book titles per capita in {default_country} for years 1948–1996",
    )
    return fig


# ================================================================
# 11) Callback pro třetí stránku – bar graf + slider (ranking populace)
# ================================================================
@app.callback(
    Output("population-bar", "figure"),
    Output("slider-help", "children"),
    Input("country-count-slider", "value"),
)
def update_population_bar(count):
    """
    Slider vrací počet zemí, které chceme zobrazit v bar grafu.
    Pomocí .head(count) vybereme N nejlidnatějších zemí.
    """
    if count is None:
        count = 10
    n = int(count)

    # vybereme prvních n zemí podle populace v roce 1996
    top_countries = selected_pops_countries.head(n)

    fig = px.bar(
        top_countries,
        x="Entity",
        y="Population (historical estimates)",
        template="none",
        title=f"Top {n} countries by population in 1996",
    )
    fig.update_layout(
        xaxis_title="Country",
        yaxis_title="Population (historical estimates)",
        xaxis_tickangle=-45,
    )

    msg = f"Zobrazuji {n} zemí podle populace v roce 1996."
    return fig, msg


# ================================================================
# 12) Spuštění serveru v samostatném vlákně + vložení do iframe
# ================================================================
PORT = 8050  # port, na kterém poběží Dash aplikace

# spuštění Dash aplikace v jiném vlákně, aby Colab neumřel
thread = threading.Thread(
    target=app.run,
    kwargs={"port": PORT, "debug": False, "use_reloader": False},
)
thread.start()

# vykreslení aplikace přímo v buňce jako iframe
output.serve_kernel_port_as_iframe(port=PORT, height=600)

# textový souhrn pod aplikací
print("\n--- Souhrn ---")
print(f"1) Dash multipage aplikace běží na portu {PORT}.")
print(f"2) Odkaz localhost (mimo Colab prostředí): http://127.0.0.1:{PORT}/")
print("3) V tomto notebooku vidíš aplikaci nad tímto textem v iframe.")
print("4) Pokud iframe nevidíš, spusť buňku znovu nebo obnov stránku.")
```
