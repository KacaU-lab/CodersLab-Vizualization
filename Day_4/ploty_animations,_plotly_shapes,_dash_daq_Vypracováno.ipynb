```python
population = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Vis/population.csv')
new_books = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Vis/new-books-per-million.csv")
data = pd.merge(new_books, population, how="inner", on=["Entity","Year"])
data = data[(data["Year"]>=1980) & (data["Year"]<=1996)]
countries_list = data[["Entity", "Year"]].groupby("Entity").count().sort_values(by="Year", ascending=False)
countries_list=countries_list[countries_list["Year"]==countries_list["Year"].max()]
data = pd.merge(data, countries_list, how="inner", on=["Entity"], suffixes=("","_z"))
data = data[data["Entity"]!="Russia"]
```

```python
from dash import Dash, dcc, html, callback, Output, Input
import dash_daq as daq
import threading
from google.colab import output

# ---------- APLIKACE ----------

app = Dash()

app.layout = html.Div(
    [
        # BooleanSwitch – bude řízen hodnotou slideru
        daq.BooleanSwitch(id='bswitch', on=False),

        # Text s informací o stavu a hodnotách ovladačů
        html.Div(id='bswitch-info'),

        # Knob (jen pro zobrazení v textu, na logiku switche nemá vliv)
        daq.Knob(id='knob', value=5),

        # Gauge + graduated bar – ukazují aktuální hodnotu slideru
        daq.Gauge(id='gauge', label='Gauge', value=8),
        daq.GraduatedBar(id='gbar', label='Graduated bar', value=6),

        # Slider – hlavní ovladač, podle něj se řídí BooleanSwitch
        html.Div(
            dcc.Slider(id='slider', min=0, max=10, step=1, value=3),
            style={"width": "100%"},
        ),
    ],
    style={
        "display": "flex",
        "flex-direction": "column",
        "justify-content": "center",
        "align-items": "center",
        "height": "100%",
        "width": "75%",
        "fontFamily": "verdana",
        "color": "#444",
    },
)

# ---------- CALLBACK 1 ----------
# Řídí BooleanSwitch i informační text
# Podmínka: pokud je slider < 5 → switch = False

@app.callback(
    Output('bswitch', 'on'),
    Output('bswitch-info', 'children'),
    Input('slider', 'value'),
    Input('knob', 'value'),
)
def update_switch(slider_value, knob_value):
    # BooleanSwitch je True jen pokud je slider >= 5
    switch_on = slider_value >= 5

    info_text = (
        f"Switched {switch_on}. "
        f"Slider is set to {slider_value}, the knob is set to {knob_value}."
    )

    return switch_on, info_text


# ---------- CALLBACK 2 ----------
# Synchronizuje Gauge a GraduatedBar se sliderem

@app.callback(
    Output('gauge', 'value'),
    Output('gbar', 'value'),
    Input('slider', 'value'),
)
def update_output(value):
    return value, value
```

Změňte způsob fungování bodového grafu tak, aby byl vybrán obdélníkový tvar. Chcete-li to provést, změňte aktuální typehodnotu na rect.

```python
import numpy as np
import pandas as pd
import plotly.express as px

# 1) Vytvoříme náhodná data
numbers = pd.DataFrame(np.random.randn(35), columns=["sample"])

# Funkce na označení max, min a median
def find_special_points(value, points):
    if value == points.max():
        special_point_name = "max"
    elif value == points.min():
        special_point_name = "min"
    elif value == points.median():
        special_point_name = "median"
    else:
        special_point_name = "not special"
    return special_point_name

numbers["special_points"] = numbers["sample"].apply(
    lambda x: find_special_points(x, numbers["sample"])
)

# 2) Základní scatter graf
chart = px.scatter(
    numbers,
    x=numbers.index,
    y="sample",
    color="special_points",
    color_discrete_map={
        "max": "#E84861",
        "median": "#169873",
        "min": "#1C4097",
        "not special": "#E4DFDA",
    },
    render_mode="webgl",
    template="none",
    size=[10 for _ in numbers["sample"]],
    title="Here are your points",
)

# 3) Definice oblasti „clusteru“ – TEĎ JAKO OBDÉLNÍK (rect)
cluster = [
    {
        "type": "rect",                 # <--- změna z 'circle' na 'rect'
        "x0": min(numbers.index),
        "x1": max(numbers.index) + 0.5,
        "y0": min(numbers["sample"]) - 1,
        "y1": max(numbers["sample"]) / 2,
        "line": {"color": "DarkOrange", "width": 2, "dash": "dashdot"},
        "fillcolor": "LightSkyBlue",
        "opacity": 0.3,
    }
]

# 4) Tlačítka: None = bez tvaru, Cluster = zobrazí obdélník
chart.update_layout(
    updatemenus=[
        dict(
            type="buttons",
            buttons=[
                dict(
                    label="None",
                    method="relayout",
                    args=["shapes", []],       # žádné tvary
                ),
                dict(
                    label="Cluster",
                    method="relayout",
                    args=["shapes", cluster],  # náš obdélník
                ),
            ],
        )
    ]
)

chart
```
