```python
app = dash.Dash()
app.layout = html.Div(
    children=[
              dcc.Graph(id='graph', figure=chart),
              dcc.Tooltip(id="tooltip"),
              ],
)

@app.callback(
    Output("tooltip", "show"),
    Output("tooltip", "bbox"),
    Output("tooltip", "children"),
    Input("graph", "hoverData")
)
def display_tooltip(data):

  if data is None:
    return False, no_update, no_update

  else:
    hovered_point = data["points"][0]
    bbox = hovered_point["bbox"]
    children = [
                html.H3("I am just showing additional data"),
                html.P(numbers.iloc[hovered_point["x"]]["description"]),
                ]

  return True, bbox, children
```

```python
app = dash.Dash()
app.layout = html.Div(
    children=[
              dcc.Graph(id='graph', figure=chart),
              dcc.Tooltip(id="tooltip"),
              ],
)

@app.callback(
    Output("tooltip", "show"),
    Output("tooltip", "bbox"),
    Output("tooltip", "children"),
    Input("graph", "hoverData"),
)
def display_tooltip(data):
    if data is None:
        # nic neukazujeme, tooltip zůstává schovaný
        return False, no_update, no_update

    # informace o právě „hovernutém“ bodu
    hovered_point = data["points"][0]
    bbox = hovered_point["bbox"]

    # fragment textu pro daný bod
    fragment = numbers.iloc[hovered_point["x"]]["description"]

    # počet slov ve fragmentu (split podle whitespace)
    word_count = len(fragment.split())

    # obsah tooltipu – místo textu fragmentu jen informace o počtu slov
    children = [
        html.H3("Informace o fragmentu"),
        html.P(f"Počet slov v tomto fragmentu: {word_count}"),
    ]

    return True, bbox, children
```

```python
thread = threading.Thread(target=app.run_server)
thread.start()
output.serve_kernel_port_as_window(port=8050)
```

```python
@app.callback(
    Output("tooltip", "show"),
    Output("tooltip", "bbox"),
    Output("tooltip", "children"),
    Input("graph", "hoverData"),
)
def display_tooltip(data):
    if data is None:
        return False, no_update, no_update

    hovered_point = data["points"][0]
    bbox = hovered_point["bbox"]
    fragment = numbers.iloc[hovered_point["x"]]["description"]

    # vytvoříme MALÝ scatter graf (kopii)
    mini_chart = px.scatter(
        numbers,
        x=numbers.index,
        y="sample",
        template="none",
        size=[8 for _ in numbers["sample"]],
        height=250,           # tooltip musí být malý
        width=350,
        title="Mini kopie grafu"
    )
    mini_chart.update_traces(hoverinfo="none", hovertemplate=None)

    # obsah tooltipu: text + graf
    children = [
        html.H4("Fragment Lorem Ipsum:"),
        html.P(fragment),
        html.Br(),
        html.H5("Kopie bodového grafu:"),
        dcc.Graph(
            figure=mini_chart,
            style={"height": "250px", "width": "350px"}
        )
    ]

    return True, bbox, children
```

```python
from google.colab import output
import threading

PORT = 8050  # můžeš změnit, když bude obsazený

def run_app():
    # nový způsob – app.run místo app.run_server
    app.run(
        host="0.0.0.0",
        port=PORT,
        debug=False,
        use_reloader=False,
    )

thread = threading.Thread(target=run_app)
thread.start()

# zobrazí Dash aplikaci přímo v buňce jako iframe
output.serve_kernel_port_as_iframe(port=PORT, height=600)

print("\n--- Souhrn ---")
print(f"1) Dash aplikace běží na portu {PORT}.")
print("2) Nad tímto textem vidíš vložené okno (iframe) s aplikací.")
print("3) Pokud iframe nevidíš, zkus buňku spustit znovu nebo obnovit stránku.")
```
